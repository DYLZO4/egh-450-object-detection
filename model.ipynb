```python
from ultralytics import YOLO
```

```python
model = YOLO('yolo11n.pt')  # or yolo11s.pt for more accuracy, slower inference
```

```python
model.train(
    data='dataset/data.yaml',
    epochs=200,
    patience=50,      # stop early if val mAP hasn't improved for 50 epochs
    imgsz=640,
    batch=16,
    cos_lr=True,      # cosine LR decay, smoother convergence over a long run
    close_mosaic=15,  # train the last 15 epochs on un-mosaiced images
    seed=42,
    name='robust',
    # colour / lighting
    hsv_h=0.015,      # hue jitter
    hsv_s=0.7,        # saturation jitter
    hsv_v=0.5,        # brightness jitter
    # geometry: downward-facing camera, so any orientation is valid
    degrees=15,       # small rotations; large ones inflate boxes and hurt mAP50-95
    flipud=0.5,       # vertical flip (probability, 1.0 would flip every image)
    fliplr=0.5,       # horizontal flip
    scale=0.6,        # zoom in/out to cover different camera heights
    translate=0.1,    # shift
    mosaic=1.0,       # combines 4 images into one during training
)
```

```python
# Evaluate on the held-out test split, alongside earlier runs for comparison
for weights in ['runs/detect/robust/weights/best.pt', 'runs/detect/train-2/weights/best.pt']:
    metrics = YOLO(weights).val(data='dataset/data.yaml', split='test', plots=False)
    print(f'{weights}: mAP50={metrics.box.map50:.3f}  mAP50-95={metrics.box.map:.3f}')
```
